## 3. Soma de dois números

Crie uma função que retorne a soma de dois números.

```python
def somar (k, j):
    c = k + j
    return(c)
```

## 5. Par ou ímpar

Retorne se o número é par ou ímpar.

```python
def parouimpar(z):
    if z%2 == 0:
        print("Seu número é par")
    else:
        print("Seu número é impar")
```

```python
parouimpar(13)
```

## 6. Maior número

Retorne o maior entre três números.

```python
def maiornumero(h, i, j):
    l = max(h, i, j)
    return(l)
    
```

```python
maiornumero(13, 33, 43)
```

## 8. Tabuada

Imprima a tabuada de um número.

```python
def tabuada(p):
    print(f"A tabuada do {p}")
    for i in range(1, 11):
        resultado = p * i
        print(f"{p} x {i} = {resultado}")
```

```python
tabuada(9)
```

## 9. Média de lista

Calcule a média de uma lista.

```python
def medialista(lista):
    if len(lista) == 0:
        return 0
    return sum(lista) / len(lista)
```

```python
numeros = [12 ,14, 74, 23]
```

```python
medialista(numeros)
```

## 10. Cadastro

Peça nome e idade e retorne um dicionário.

```python
def cadastrar_pessoa():
    nome = input("Digite o seu nome: ")
    idade = int(input("Digite a sua idade: "))
    
    usuario = {
        "nome": nome,
        "idade": idade}
    
    return usuario
```

```python
cadastrar_pessoa()
```
